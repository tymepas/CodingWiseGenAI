# Conversational AI

```python
max_conversations = 20
conversational_length = 0
```

```python
message_history = [
    {'role':'system','content': system_message},
    {'role':'user','content':'help me solve the equation 3x-9=21'},
    {'role':'assistant','content':'Sure! Try moving 9 towards the right hand side of equation. What do you get'},
    {'role':'user','content':'3x=12+21'},
    {'role':'assistant','content':'Well! There seems an error in your calculations. When you move 9 to the right it becomes 3x=9+21 not 3x=12+21'},
    {'role':'user','content':'3x=30'},
    {'role':'assistant','content':'Hey Looks like a great job! Now try to divide both sides by 3. What do you get?'},
    {'role':'user','content':'x=10'},
    {'role':'user','content':'Help me to sole the equation x-10=2x'}
    ]
```

```python
while conversational_length < max_conversations:
  user_input = input('You: ')

  if 'exit' in user_input.lower():
    print("AI Tutor: Exiting the application")
    break
  message_history.append({'role': 'user', 'content': user_input})

  chat_response = openai.chat.completions.create(
      model='gpt-3.5-turbo-16k',
      messages=message_history,
      max_tokens=1000,
      temperature=0.5,
      n=1,
      stop=None,
      frequency_penalty=0,
      presence_penalty=0
  )

  ai_response = chat_response.choices[0].message.content
  print(f'AI Tutor: {ai_response}')
  print("\n")
  message_history.append({'role': 'assistant', 'content': ai_response})
  conversational_length += 1
```

```python
# check tokens, how much we have spent

def chat_response_with_num_tokens(messages):
  chat_response = openai.chat.completions.create(
      model='gpt-3.5-turbo-16k',
      messages=messages
  )
  content = chat_response.choices[0].message.content
  tokens_count = {
      'prompt_tokens': chat_response.usage.prompt_tokens,
      'completion_tokens': chat_response.usage.completion_tokens,
      'total_tokens': chat_response.usage.total_tokens
  }
  return content, tokens_count
```

```python
def chat_response_with_num_tokens(messages):
    chat_response = openai.chat.completions.create(
        model='gpt-3.5-turbo-16k',
        messages=messages,
        max_tokens=1000,
        temperature=0.5,
        n=1,
        stop=None,
        frequency_penalty=0,
        presence_penalty=0
    )
    content = chat_response.choices[0].message.content
    tokens_count = {
        'prompt_tokens': chat_response.usage.prompt_tokens,
        'completion_tokens': chat_response.usage.completion_tokens,
        'total_tokens': chat_response.usage.total_tokens
    }
    return content, tokens_count


total_tokens_used = 0

while conversational_length < max_conversations:
    user_input = input('You: ')

    if 'exit' in user_input.lower():
        print("AI Tutor: Exiting the application")
        break

    message_history.append({'role': 'user', 'content': user_input})

    ai_response, tokens_count = chat_response_with_num_tokens(message_history)

    print(f'AI Tutor: {ai_response}')
    print(f"[tokens] prompt: {tokens_count['prompt_tokens']} | "
          f"completion: {tokens_count['completion_tokens']} | "
          f"turn total: {tokens_count['total_tokens']}")

    total_tokens_used += tokens_count['total_tokens']
    print(f"[running total this session: {total_tokens_used} tokens]\n")

    message_history.append({'role': 'assistant', 'content': ai_response})
    conversational_length += 1
```
