# **Summarziation**

```python
summary = pipeline("summarization")
```

```python
text = """Paste a longer paragraph here — a few sentences minimum,
summarization needs enough input to actually compress."""
result = summary(text)
print(result[0]['summary_text'])
```

```python
text = """The Amazon rainforest, often referred to as the lungs of the Earth,
spans over 5.5 million square kilometers across nine countries in South America.
It is home to an estimated 400 billion individual trees representing 16,000
species, and roughly 10% of the world's known species live within its boundaries.
Deforestation driven by agriculture, logging, and mining has reduced forest
cover significantly over the past 50 years, raising serious concerns about
biodiversity loss and the forest's role in regulating global carbon levels."""

result = summary(text, max_length=60, min_length=20)
print(result[0]['summary_text'])
```

```python
summary("""You want to understand how your users behave. Specifically: what is the typical session duration? What fraction convert? What do they actually think of the new onboarding flow?

The honest answer requires data from every user. That's the ground truth. But your product has millions of users, surveys have response costs, and querying your full event log for every analysis would be slow and expensive. Measuring everyone for every question is impractical. So you measure a subset and use it to make claims about the whole. This is sampling. The question isn't whether to sample — you nearly always are, whether you realize it or not. The question is whether your sample will give you accurate information about the population it's meant to represent.

The natural first approach is to use whoever is convenient. Survey users who open the email. Analyze the first thousand signups. Pull data from the most active users because their behavior is easiest to observe. This fails in a specific way: convenience samples are systematically biased. Email-openers are more engaged than your typical user. The first thousand signups are early adopters — typically more forgiving and more technically sophisticated. Active users are, by definition, not representative of the majority who churn quietly. Every convenience shortcut introduces a systematic difference between your sample and the population you're claiming to understand.

A biased sample doesn't get better with size. Ten thousand email-openers are still all email-openers. More data from the wrong population just makes you more confidently wrong.

What you need is a selection mechanism where each member of the population has a known, equal probability of being chosen — independent of the outcome you're measuring. This is random sampling. The randomness breaks the link between selection and outcome. If session duration doesn't influence whether you get selected, the sample's session distribution will match the population's — not perfectly (there's always variance), but without systematic bias.

Sample size matters, but its relationship to accuracy is often misunderstood. Accuracy is determined by absolute sample size, not the fraction of the population you've captured. A sample of 1,000 from a population of 10,000 is not meaningfully more accurate than a sample of 1,000 from a population of 100 million — what matters is n, not n/N.""")
```

# **Zero-Shot Classification**

```python
zero_shot = pipeline("zero-shot-classification")
```

```python
zero_shot("the course is of Generative Ai and NLP and we are learning about hugging Face",
          candidate_labels=["education", "technology", "business", "politics", "science","food"])
```

```python
zero_shot("Tesla's new AI chip could cut manufacturing costs by 30% next quarter",
          candidate_labels=["technology", "business", "science", "politics"])
```

```python
zero_shot("Oh great, another Monday meeting that could've been an email",
          candidate_labels=["positive", "negative", "neutral"])
```

```python
zero_shot("The chef used liquid nitrogen and 3D-printed molds to plate the dessert",
          candidate_labels=["food", "technology", "art", "science"],
          multi_label=True)
```

```python
zero_shot("I can't believe they cancelled the show after one season",
          candidate_labels=["entertainment", "disappointment", "news"])
```

```python
zero_shot("यह कोर्स आर्टिफिशियल इंटेलिजेंस पर आधारित है",
          candidate_labels=["education", "technology", "sports"])
```

```python
zero_shot_multi = pipeline("zero-shot-classification", model="joeddav/xlm-roberta-large-xnli")
```

```python
zero_shot_multi("यह कोर्स आर्टिफिशियल इंटेलिजेंस पर आधारित है",
          candidate_labels=["education", "technology", "sports"])
```

# **NER = Named Entity Recognition.**



```python
ner = pipeline("ner", grouped_entities=True)
```

```python
ner("Elon Musk founded SpaceX in California in 2002")
```

```python
ner("Sundar Pichai, CEO of Google, met with officials in New Delhi last Tuesday")
```

```python
ner("Amazon reported strong sales while deforestation in the Amazon worsened")
```

```python
ner_raw = pipeline("ner")
ner_raw("Shah Rukh Khan works at Red Chillies Entertainment")
```

```python
ner("yo i just met priyanka chopra outside starbucks in mumbai lol crazy")
```

```python
text = "Please contact John Mathews at john.mathews@email.com regarding the Infosys contract"
entities = ner(text)
for e in entities:
    if e['entity_group'] == 'PER':
        text = text.replace(e['word'], '[REDACTED]')
print(text)
```

# **Questions & Answers**



```python
from transformers import pipeline

pipe = pipeline("question-answering", model="ahotrod/electra_large_discriminator_squad2_512")
```

```python
result = pipe(
    question="Who is the CEO of the company?",
    context="Infosys is an Indian IT services company. Salil Parekh has served as CEO and Managing Director since 2018, overseeing operations across 50+ countries."
)
print(result)
```

```python
result = pipe(
    question="How many records does the dataset contain?",
    context="The sales dataset covers Q3 2026 performance across all regional stores. It contains 120,000 transaction records collected between July and September, with fields for product ID, quantity, and revenue."
)
print(result)
```
